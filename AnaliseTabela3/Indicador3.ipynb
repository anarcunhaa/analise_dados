```python
import pandas as pd
```

```python
path = r"H:\Python\Etapa3\AnaliseTabela3\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"

dados = pd.read_excel(path, header=None)
```

```python
print(dados)
```

```python
# Definimos a lista com os nomes exatos que estão na coluna 0
regioes = ['Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']

# Filtramos o DataFrame original mantendo apenas as linhas onde a coluna 0 é uma região
dados_regioes = dados[dados[0].isin(regioes)].copy()
```

```python
#Tabela 1 - Homens e mulheres por região com curso superior
tabela1 = dados_regioes[[0, 1, 2, 3]].copy()
tabela1.columns = ['Região', 'Total_Geral', 'Homens_Geral', 'Mulheres_Geral']

# Converter valores para numérico para não quebrar o describe()
tabela1[['Total_Geral', 'Homens_Geral', 'Mulheres_Geral']] = tabela1[['Total_Geral', 'Homens_Geral', 'Mulheres_Geral']].apply(pd.to_numeric)

# Salvar arquivo
tabela1.to_excel("tabela_curso_superior.xlsx", index=False)

print("=== ESTATÍSTICAS DA TABELA 1 ===")
print(tabela1.describe().round(2))
print("\n" + "="*50 + "\n")
```

```python
#Tabela 2 - Homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
# Coluna 4: Total STEM
# Coluna 5: Homens STEM
# Coluna 6: Mulheres STEM

tabela2 = dados_regioes[[0, 4, 5, 6]].copy()
tabela2.columns = ['Região', 'Total_STEM', 'Homens_STEM', 'Mulheres_STEM']

# Converter valores para numérico
tabela2[['Total_STEM', 'Homens_STEM', 'Mulheres_STEM']] = tabela2[['Total_STEM', 'Homens_STEM', 'Mulheres_STEM']].apply(pd.to_numeric)

# Salvar arquivo usando o nome sugerido pelo seu exercício (.to_excel)
tabela2.to_excel("tabela_tecnologia.xlsx", index=False)

print("=== ESTATÍSTICAS DA TABELA 2 ===")
print(tabela2.describe().round(2))
print("\n" + "="*50 + "\n")
```

```python
#Tabela 3 - Homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar
# Coluna 7: Total Educação/Saúde
# Coluna 8: Homens Educação/Saúde
# Coluna 9: Mulheres Educação/Saúde

tabela3 = dados_regioes[[0, 7, 8, 9]].copy()
tabela3.columns = ['Região', 'Total_Saude_Educacao', 'Homens_Saude_Educacao', 'Mulheres_Saude_Educacao']

# Converter valores para numérico
tabela3[['Total_Saude_Educacao', 'Homens_Saude_Educacao', 'Mulheres_Saude_Educacao']] = tabela3[['Total_Saude_Educacao', 'Homens_Saude_Educacao', 'Mulheres_Saude_Educacao']].apply(pd.to_numeric)

# Salvar arquivo
tabela3.to_excel("tabela_saude_educacao.xlsx", index=False)

print("=== ESTATÍSTICAS DA TABELA 3 ===")
print(tabela3.describe().round(2))
```
